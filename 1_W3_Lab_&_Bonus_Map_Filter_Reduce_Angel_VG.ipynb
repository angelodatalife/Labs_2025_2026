# Challenge 2 - Filtering

When printing out a few words from the book, we see that there are words that we may not want to keep if we choose to analyze the corpus of text. Below is a list of words that we would like to get rid of. Create a function that will return false if it contains a word from the list of words specified and true otherwise.

```python
def word_filter(x):
    '''
    Input: A string
    Output: True if the word is not in the specified list 
    and False if the word is in the list.
        
    Example:
    word list = ['and', 'the']
    Input: 'and'
    Output: False
    
    Input: 'John'
    Output: True
    '''
    
    word_list = ['and', 'the', 'a', 'an']
    
```

Use the `filter()` function to filter out the words speficied in the `word_filter()` function. Store the filtered list in the variable `prophet_filter`.

```python
# your code here

from typing import List

# 1.  Define the words we want to get rid of
word_list: List[str] = ['and', 'the', 'a', 'an']

# 2.  Filter function: True → keep word, False → discard word
def word_filter(a_word: str) -> bool:
    """
    Return True  if the word is NOT in word_list (keep it)
    Return False if the word IS in word_list (discard it)
    """
    return a_word.lower() not in word_list

# 3.  Apply Python’s built-in filter()
#     filter() keeps only elements that return True from word_filter
prophet_filter = list(filter(word_filter, prophet_flat))

# 4.  Quick sanity check
print("Original words :", len(prophet_flat))
print("After filtering:", len(prophet_filter))
print("First 10 kept words:", prophet_filter[:10])
```

```python
# your code here

"""
Goal: turn the clean word-list back into one long string, each word separated by a space.
We use functools.reduce to apply concat_space cumulatively across the list.
"""

from functools import reduce

# 1.  Helper: glue two strings together with a space in between
def concat_space(a: str, b: str) -> str:
    """
    Input : two strings
    Output: single string separated by a space
    Example: concat_space('John', 'Smith') -> 'John Smith'
    """
    return a + ' ' + b

# 2.  Rebuild the entire corpus into one string
prophet_reduce = reduce(concat_space, prophet_filter)  # prophet_filter is the cleaned list

# 3.  Quick sanity check
print("First 200 chars of rebuilt text:")
print(prophet_reduce[:200])
```
