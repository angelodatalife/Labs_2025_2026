```python
### [your code here]
# Multiply a and c element-wise and store the result in e
e = a * c

# Print basic info
print("Shape of e:", e.shape)
print("Sample values from e:")
print(e)

# ------------------------------------------------------------------
# Since c is an array of ones with the same shape as a,
# every element in e equals the corresponding element in a multiplied by 1.
# Therefore, e is identical to a.
# In short:  e = a * 1  -->  e = a
# ------------------------------------------------------------------
```



### 13. Does e equal to a? Why or why not?


```python
### [your code here]
# Check whether e and a are numerically identical
are_equal = np.array_equal(e, a)

print("Does e equal a?", are_equal)

# ------------------------------------------------------------------
# e was obtained by element-wise multiplication of a and c.
# c is an array of ones, so each element of e is a[i,j,k] * 1.
# Multiplying by 1 does not change the value, hence e is numerically
# identical to a. Floating-point rounding errors are not expected here
# because 1.0 is exactly representable and multiplication by 1.0 is exact.
# ------------------------------------------------------------------
```

### 18. Bonus question: instead of using numbers (i.e. 0, 25, 50, 75, and 100), how to use string values 
("A", "B", "C", "D", and "E") to label the array elements? You are expecting the result to be:
array([[[ 'D',  'D',  'D',  'B',  'D'],
        [ 'D',  'D',  'B',  'B',  'B'],

```python
### [your code here]

# Create an empty array of strings with the same shape as d
f_str = np.empty(d.shape, dtype='<U1')   # '<U1' = single-character string

# Populate it with letters A-E according to the same rules
for i in range(d.shape[0]):
    for j in range(d.shape[1]):
        for k in range(d.shape[2]):
            val = d[i, j, k]
            if val == d_min:
                f_str[i, j, k] = 'A'
            elif val == d_max:
                f_str[i, j, k] = 'E'
            elif val == d_mean:
                f_str[i, j, k] = 'C'
            elif d_min < val < d_mean:
                f_str[i, j, k] = 'B'
            elif d_mean < val < d_max:
                f_str[i, j, k] = 'D'

print("f with string labels:")
print(f_str)
```
