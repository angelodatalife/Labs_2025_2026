## 2. Now write a function that returns the largest element on a list

```python
#your code here
def largest_in_list(numbers):
    if not numbers:
        return None
    
    largest = numbers[0]
    for num in numbers[1:]:
        if num > largest:
            largest = num
    return largest
```

```python
# This will test your function 
def test_largest_in_list():
    assert largest_in_list([3, 7, 2, 9, 5]) == 9
    assert largest_in_list([-1, -5, -3]) == -1
    assert largest_in_list([42]) == 42
    assert largest_in_list([]) is None
    print("All tests passed")

test_largest_in_list()
```

## 5. Now combine those two ideas and write a function that receives a list and either "+" or "*" and outputs acordingly

```python
#your code here
def oper_all(numbers, oper="*"):
    if not numbers:
        return 0 if oper == "+" else 1

    if oper == "+":
        result = 0
        for num in numbers:
            result += num
        return result
    elif oper == "*":
        result = 1
        for num in numbers:
            result *= num
        return result
    else:
        raise ValueError("Unsupported operation. Use '+' or '*'")
```

```python
# This will test your function 
def test_oper_all():
    assert oper_all([1, 2, 3], "+") == 6
    assert oper_all([2, 3, 4], "*") == 24
    assert oper_all([], "+") == 0
    assert oper_all([], "*") == 1
    assert oper_all([5], "+") == 5
    assert oper_all([5], "*") == 5
    print("All tests passed ✅")

test_oper_all()
```

## 6. Write a function that returns the factorial of a number.

```python
#your code here
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
```

```python
#factorial formula
#n! = n * ( n - 1 ) *...*1

# This code defines a function called "factorial" which takes an input "n". The function uses a for loop to iterate through the range of numbers 
# from 1 to n+1. For each number in that range, it multiplies the current value of x by the number in the range. At the end of the loop, 
# the function returns the final value of x, which will be the factorial of the input number "n".

# The Factorial of a positive integer n is the product of all positive integers less than or equal to n. 
# For example, the factorial of 6 (written "6!") is 6 * 5 * 4 * 3 * 2 * 1 = 720.

# So this function takes an input of any positive integer, and returns the factorial of that number.
```

```python
# This will test your function 
def test_factorial():
    assert factorial(0) == 1
    assert factorial(1) == 1
    assert factorial(5) == 120
    assert factorial(7) == 5040
    print("All tests passed ✅")

# Run the test
test_factorial()
```

## 7. Write a function that takes a list and returns a list of the unique values.

`NOTE: You cannot use set. 🤔`

```python
#your code here
def unique(list):
    uniques = []
    for item in list:
        if item not in uniques:
            uniques.append(item)
    return uniques
```

```python
# This will test your function 
def test_unique():
    assert unique([1, 2, 2, 3, 1, 4]) == [1, 2, 3, 4]
    assert unique([]) == []
    assert unique([7, 7, 7]) == [7]
    assert unique(["a", "b", "a"]) == ["a", "b"]
    print("All tests passed ✅")

# Run the test
test_unique()
```

## 8. Write a function that returns the mode of a list, i.e.: the element that appears the most times.
`NOTE: You should not use count... 🧐`

```python
#your code here
def mode_counter(arr):
    if not arr:
        return None  # or raise ValueError("Empty list has no mode")

    freq = {}          # 1. empty dictionary: {element: times}
    for item in arr:   # 2. count manually
        if item in freq:
            freq[item] += 1
        else:
            freq[item] = 1

    # 3. find key with max value
    max_val = 0
    mode = None
    for key, value in freq.items():
        if value > max_val:
            max_val = value
            mode = key
    return mode
```

```python
# This will test your function 
def test_mode_counter():
    assert mode_counter([1, 2, 2, 3, 2]) == 2
    assert mode_counter([7, 7, 7, 8, 8]) == 7
    assert mode_counter([5]) == 5
    assert mode_counter([]) is None
    assert mode_counter(["a", "b", "b", "c"]) == "b"
    print("All tests passed ✅")

# Run the test
test_mode_counter()
```

## 9. Write a function that calculates the standard deviation of a list.
`NOTE: Do not use any libraries or already built functions. 😉`

```python
#your code here
def st_dev(list_sd):
    if len(list_sd) == 0:
        return None  # or raise ValueError("Empty list")

    n = len(list_sd)

    # 1. Calculate mean
    mean = sum(list_sd) / n

    # 2. Calculate squared differences
    squared_diffs = []
    for x in list_sd:
        diff = x - mean
        squared_diffs.append(diff ** 2)

    # 3. Calculate variance
    variance = sum(squared_diffs) / n

    # 4. Return standard deviation (square root)
    return variance ** 0.5
```

```python
# This will test your function 
def test_st_dev():
    assert round(st_dev([10, 10, 10]), 2) == 0.0
    assert round(st_dev([2, 4, 4, 4, 5, 5, 7, 9]), 2) == 2.0
    assert round(st_dev([1, 2, 3, 4, 5]), 2) == 1.41
    print("All tests passed ✅")

# Run the test
test_st_dev()
```

## 10. Write a function to check if a string is a pangram, i.e.: if it contains all the letters of the alphabet at least once. Mind that the strings may contain characters that are not letters.

```python
#your code here
def pangram(text):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    clean = text.lower()          # ignore case
    for letter in alphabet:
        if letter not in clean:   # missing any letter?
            return False
    return True
```

```python
# This will test your function 
print(pangram("abcdefghijklmnopqrstuvwxyz"))
print(pangram("abc"))
```

## 11. Write a function that receives a string of comma separated words and returns a string of comma separated words sorted alphabetically.

`NOTE: You may use sorted but not split and definitely no join! 🤪`

```python
#your code here
def sort_alpha(string):
    # 1. manual split by comma
    words = []
    current = ""
    for ch in string:
        if ch == ",":
            words.append(current)
            current = ""
        else:
            current += ch
    if current:               # last word after final comma
        words.append(current)

    # 2. sort alphabetically
    words = sorted(words)

    # 3. manual join with commas
    result = ""
    for i, w in enumerate(words):
        result += w
        if i != len(words) - 1:
            result += ","
    return result
```

```python
# This will test your function 
def test_sort_alpha():
    original = "banana,apple,cherry,date"
    expected = "apple,banana,cherry,date"
    print("Original :", original)
    print("Sorted   :", sort_alpha(original))
    print("Expected :", expected)

# Run the test
test_sort_alpha()
```

## 12. Write a function to check if a given password is strong (at least 8 characters, at least one lower case, at least one upper case, at least one number and at least one special character). It should output True if strong and False if not.

```python
#your code here
def check_pass(password):
    if len(password) < 8:
        return False

    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False
    special_chars = "!@#$%^&*()-_+="

    for ch in password:
        if ch.islower():
            has_lower = True
        elif ch.isupper():
            has_upper = True
        elif ch.isdigit():
            has_digit = True
        elif ch in special_chars:
            has_special = True

    return has_lower and has_upper and has_digit and has_special
```

```python
# This will test your function 
def test_check_pass():
    passwords = ["weak", "Weak123", "Weak123!", "Strong123!", "NOLOWER123!", "noupper123!", "NoSpecial123"]
    for pwd in passwords:
        print(f"{pwd:15} -> {check_pass(pwd)}")

# Run the test
test_check_pass()
```
