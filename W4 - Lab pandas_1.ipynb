```python
# your code here

# 1. Import pandas
import pandas as pd

# 2. Build the dictionary with cleaned keys and lists
orders = {
    'Description': [
        'LUNCH BAG APPLE DESIGN',
        'SET OF 60 VINTAGE LEAF CAKE CASES',
        'RIBBON REEL STRIPES DESIGN',
        'WORLD WAR 2 GLIDERS ASSTD DESIGNS',
        'PLAYING CARDS JUBILEE UNION JACK',
        'POPCORN HOLDER',
        'BOX OF VINTAGE ALPHABET BLOCKS',
        'PARTY BUNTING',
        'JAZZ HEARTS ADDRESS BOOK',
        'SET OF 4 SANTA PLACE SETTINGS'
    ],
    'Quantity': [1, 24, 1, 2880, 2, 7, 1, 4, 10, 48],
    'UnitPrice': [1.65, 0.55, 1.65, 0.18, 1.25, 0.85, 11.95, 4.95, 0.19, 1.25],
    'Revenue': [1.65, 13.2, 1.65, 518.4, 2.5, 5.95, 11.95, 19.8, 1.9, 60.0]
}

# 3. Create the DataFrame
df_orders = pd.DataFrame(orders)

# 4. Display the DataFrame
print(df_orders)
```

### 11. Calculate the total quantity ordered and revenue generated from these orders.

```python
# your code here

# 1. Compute total quantity and total revenue
total_qty   = df_orders['Quantity'].sum()
total_revenue = df_orders['Revenue'].sum()

# 2. Display results
print("Total quantity ordered :", total_qty)
print("Total revenue generated:", total_revenue)
```

### 12. Obtain the prices of the most expensive and least expensive items ordered and print the difference.

```python
# your code here

# 1. Compute the price difference for each order
df_orders['PriceDiff'] = df_orders['UnitPrice'].diff()

# 2. Display the price differences
print(df_orders[['Description', 'UnitPrice', 'PriceDiff']])
```

### 3 - In this part of the lab, we would like to test complex conditions on the entire data set at once. Let's start by finding the number of rows where the CGPA is greater than 9 and the student has performed an investigation.

```python
# your code here

# Apply the complex condition
count = admissions[(admissions['CGPA'] > 9) & (admissions['Research'] == 1)].shape[0]

print('Number of students with CGPA > 9 and research experience:', count)
```

### 4 - Now return all the rows where the CGPA is greater than 9 and the SOP score is less than 3.5. Find the mean chance of admit for these applicants.

```python
# your code here

# 1. Filter rows: CGPA > 9  &  SOP < 3.5
high_cgpa_low_sop = admissions[(admissions['CGPA'] > 9) & (admissions['SOP'] < 3.5)]

# 2. Compute mean chance of admit for these applicants
mean_chance = high_cgpa_low_sop['Chance of Admit '].mean()

# 3. Display the result
print(f'Rows with CGPA > 9 and SOP < 3.5:')
print(high_cgpa_low_sop)

print(f'\nMean chance of admit for these applicants: {mean_chance:.3f}')
```
